````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
DB_PATH = "club.db"
EVENTS = ["迎新晚會", "程式工作坊", "籃球友誼賽"]
INITIAL_EVENTS = [
    ("迎新晚會", "2026-11-20", 6),
    ("程式工作坊", "2026-11-25", 4),
    ("籃球友誼賽", "2026-11-28", 3),
]
def get_conn():
    return sqlite3.connect(DB_PATH)
def init_db():
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("DROP TABLE IF EXISTS registrations")
        cur.execute("DROP TABLE IF EXISTS events")
        cur.execute("""
            CREATE TABLE events (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL UNIQUE,
                date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)
        cur.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student TEXT NOT NULL,
                event_id INTEGER NOT NULL,
                headcount INTEGER NOT NULL,
                status TEXT NOT NULL DEFAULT '已報名',
                FOREIGN KEY (event_id) REFERENCES events(id)
            )
        """)
        cur.executemany(
            "INSERT INTO events (name, date, capacity) VALUES (?, ?, ?)",
            INITIAL_EVENTS,
        )
        conn.commit()
    finally:
        conn.close()
init_db()
print("✅ 資料庫已重建，初始資料已放入")
````

````python
def show_tables():
    conn = get_conn()
    try:
        print("📌 events（活動）")
        display(pd.read_sql_query("SELECT * FROM events", conn))
        print("📌 registrations（報名紀錄）")
        display(pd.read_sql_query("SELECT * FROM registrations", conn))
    finally:
        conn.close()
show_tables()
````

````python
ACTIONS = ["報名", "取消", "查詢", "統計"]
MAX_RETRY = 3
SYSTEM_PROMPT = """你是系學會活動報名系統的「句子解讀器」。
請從使用者的一句話中，抽出三個欄位，並只輸出一行：動作,活動,人數
規則：
1. 動作只能是：報名、取消、查詢、統計。
   - 想參加、報名、登記 → 報名
   - 不去了、退出、取消 → 取消
   - 還有名額嗎、剩幾個位子 → 查詢
   - 各活動報名狀況、所有活動統計、報名人數總覽 → 統計
2. 活動只能是：迎新晚會、程式工作坊、籃球友誼賽。沒提到或不是這三個，輸出 無。
3. 人數是「這次要報名的總人數，包含說話的人自己」。
   例如「我和兩個朋友」= 3，「我們兩個」= 2，沒說人數 = 1。只輸出阿拉伯數字。
4. 完全無法判斷時輸出：未知,無,1
5. 只輸出一行「動作,活動,人數」，不要解釋、不要加其他文字。
你不可以：回答問題、執行報名、編造活動或人數。"""
_ai_cache = {}
def call_groq(text):
    for attempt in range(MAX_RETRY + 1):
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=0,
                reasoning_effort=REASONING_EFFORT,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": text},
                ],
            )
            time.sleep(2)
            return resp.choices[0].message.content or ""
        except Exception as e:
            if "429" in str(e) and attempt < MAX_RETRY:
                print("⚠️ 超過使用量，30 秒後重試（第", attempt + 1, "次）")
                time.sleep(30)
                continue
            print("⚠️ AI 呼叫失敗：", type(e).__name__)
            return ""
    return ""
def parse_reply(reply):
    action, pos = "未知", None
    for a in ACTIONS:
        i = reply.find(a)
        if i != -1 and (pos is None or i < pos):
            action, pos = a, i
    activity, pos = "未知", None
    for ev in EVENTS:
        i = reply.find(ev)
        if i != -1 and (pos is None or i < pos):
            activity, pos = ev, i
    rest = reply
    for w in ACTIONS + EVENTS:
        rest = rest.replace(w, "")
    nums = re.findall(r"\d+", rest)
    count = int(nums[0]) if nums else 1
    return action, activity, count
def interpret(text):
    text = (text or "").strip()
    if not text:
        return "未知", "未知", 1, ""
    if text not in _ai_cache:
        _ai_cache[text] = call_groq(text)
    raw = _ai_cache[text]
    action, activity, count = parse_reply(raw)
    return action, activity, count, raw
print("✅ AI 解讀函式已定義")
````

````python
AI_TESTS = [
    ("我想參加迎新晚會", "報名", "迎新晚會", 1),
    ("幫我取消程式工作坊", "取消", "程式工作坊", "*"),
    ("籃球友誼賽還有名額嗎", "查詢", "籃球友誼賽", "*"),
    ("我和兩個朋友想報名籃球友誼賽", "報名", "籃球友誼賽", 3),
    ("各活動報名狀況", "統計", "*", "*"),
    ("今天天氣如何", "未知", "*", "*"),
    ("我要報名羽球比賽", "*", "未知", "*"),
    ("我帶五個朋友報名迎新晚會", "報名", "迎新晚會", 6),
    ("我想報名", "報名", "未知", "*"),
    ("我和同學一共兩個人報名程式工作坊", "報名", "程式工作坊", 2),
]
def run_ai_tests(tests):
    passed = 0
    for i, (text, exp_a, exp_e, exp_c) in enumerate(tests, 1):
        a, e, c, raw = interpret(text)
        ok = (exp_a == "*" or a == exp_a) and (exp_e == "*" or e == exp_e) and (exp_c == "*" or c == exp_c)
        passed += ok
        print(f"{i}. {'✅' if ok else '❌'} 「{text}」")
        print(f"   AI 原始回覆：{raw!r}")
        print(f"   系統解讀：動作={a}，活動={e}，人數={c}　（預期：{exp_a}，{exp_e}，{exp_c}）")
    print(f"\n🎯 AI 解讀測試：通過 {passed} / {len(tests)} 句")
run_ai_tests(AI_TESTS)
````

````python
MAX_PER_PERSON = 2
MIN_COUNT, MAX_COUNT = 1, 3
def get_event(cur, name):
    cur.execute("SELECT id, name, date, capacity FROM events WHERE name = ?", (name,))
    return cur.fetchone()
def get_remaining(cur, event_id, capacity):
    cur.execute(
        "SELECT COALESCE(SUM(headcount), 0) FROM registrations WHERE event_id = ? AND status = '已報名'",
        (event_id,),
    )
    row = cur.fetchone()
    used = row[0] if row is not None else 0
    return capacity - used
def list_events():
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("SELECT name, date FROM events ORDER BY id")
        return "、".join(f"{n}（{d}）" for n, d in cur.fetchall())
    finally:
        conn.close()
def do_register(student, event_name, count):
    if count < MIN_COUNT or count > MAX_COUNT:
        return f"❌ 報名人數需在 {MIN_COUNT}～{MAX_COUNT} 人之間，你說的是 {count} 人"
    conn = get_conn()
    try:
        cur = conn.cursor()
        ev = get_event(cur, event_name)
        if ev is None:
            return "❌ 找不到這個活動"
        event_id, name, date, capacity = ev
        cur.execute(
            "SELECT id FROM registrations WHERE student = ? AND event_id = ? AND status = '已報名'",
            (student, event_id),
        )
        if cur.fetchone() is not None:
            return f"❌ {student} 已經報名過「{name}」了"
        cur.execute(
            "SELECT COUNT(*) FROM registrations WHERE student = ? AND status = '已報名'",
            (student,),
        )
        row = cur.fetchone()
        active = row[0] if row is not None else 0
        if active >= MAX_PER_PERSON:
            return f"❌ {student} 已報名 {active} 個活動，達到每人 {MAX_PER_PERSON} 個的上限"
        remaining = get_remaining(cur, event_id, capacity)
        if remaining < count:
            return f"❌ 「{name}」名額不足：剩餘 {remaining} 人，你要報 {count} 人"
        cur.execute(
            "INSERT INTO registrations (student, event_id, headcount, status) VALUES (?, ?, ?, '已報名')",
            (student, event_id, count),
        )
        conn.commit()
        return f"✅ 報名成功！{student} 報名「{name}」{count} 人，剩餘名額 {remaining - count} 人"
    finally:
        conn.close()
def do_cancel(student, event_name):
    conn = get_conn()
    try:
        cur = conn.cursor()
        ev = get_event(cur, event_name)
        if ev is None:
            return "❌ 找不到這個活動"
        event_id, name, date, capacity = ev
        cur.execute(
            "SELECT id, headcount FROM registrations WHERE student = ? AND event_id = ? AND status = '已報名'",
            (student, event_id),
        )
        rec = cur.fetchone()
        if rec is None:
            return f"❌ 查無 {student} 報名「{name}」的紀錄，無法取消"
        cur.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (rec[0],))
        conn.commit()
        remaining = get_remaining(cur, event_id, capacity)
        return f"✅ 已取消 {student} 的「{name}」報名（釋出 {rec[1]} 人），剩餘名額 {remaining} 人"
    finally:
        conn.close()
def do_query(event_name):
    conn = get_conn()
    try:
        cur = conn.cursor()
        ev = get_event(cur, event_name)
        if ev is None:
            return "❌ 找不到這個活動"
        event_id, name, date, capacity = ev
        remaining = get_remaining(cur, event_id, capacity)
        return f"📋 「{name}」（{date}）名額 {capacity} 人，剩餘 {remaining} 人"
    finally:
        conn.close()
def do_stats():
    conn = get_conn()
    try:
        cur = conn.cursor()
        cur.execute("""
            SELECT e.name, e.date, e.capacity, COALESCE(SUM(r.headcount), 0) AS registered
            FROM events e
            LEFT JOIN registrations r ON r.event_id = e.id AND r.status = '已報名'
            GROUP BY e.id, e.name, e.date, e.capacity
            ORDER BY e.id
        """)
        lines = ["📊 各活動報名狀況"]
        for name, date, capacity, registered in cur.fetchall():
            lines.append(f"• {name}（{date}）：名額 {capacity}，已報名 {registered}，剩餘 {capacity - registered}")
        return "\n".join(lines)
    finally:
        conn.close()
print("✅ 功能函式已定義")
````

````python
def handle_request(text, student):
    student = (student or "").strip()
    action, activity, count, raw = interpret(text)
    if action == "統計":
        return do_stats()
    if action == "未知" or activity == "未知":
        return "😕 我看不懂你想做什麼或哪個活動。目前的活動有：" + list_events()
    if action == "查詢":
        return do_query(activity)
    if not student:
        return "❌ 報名與取消需要輸入姓名"
    if action == "報名":
        return do_register(student, activity, count)
    if action == "取消":
        return do_cancel(student, activity)
    return "😕 無法處理，目前的活動有：" + list_events()
init_db()
FLOW_TESTS = [
    ("小明", "我想參加迎新晚會", "報名成功"),
    ("小明", "我想參加迎新晚會", "已經報名"),
    ("小明", "我和兩個朋友想報名籃球友誼賽", "報名成功"),
    ("小明", "我想報名程式工作坊", "上限"),
    ("小華", "我想報名籃球友誼賽", "名額不足"),
    ("小明", "幫我取消籃球友誼賽", "已取消"),
    ("小華", "幫我取消程式工作坊", "查無"),
    ("小華", "我帶四個朋友報名程式工作坊", "人數"),
    ("", "程式工作坊還有名額嗎", "剩餘 4"),
    ("", "各活動報名狀況", "報名狀況"),
]
passed = 0
for i, (who, text, keyword) in enumerate(FLOW_TESTS, 1):
    reply = handle_request(text, who)
    ok = keyword in reply
    passed += ok
    print(f"{i}. {'✅' if ok else '❌'} [{who or '（無姓名）'}] {text}")
    print("   →", reply.replace("\n", "\n     "))
print(f"\n🎯 流程測試：通過 {passed} / {len(FLOW_TESTS)} 步\n")
show_tables()
````
